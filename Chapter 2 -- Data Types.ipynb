# 7. Dictionary: similar to list, but order doesn't matter and data structure is based on key-value pairs
* Keys are unique. If same keys are provided, the latter value of the keys will overwrite previous value(s)

```python
a = {
    'c': 1,
    'b': 'something',
    'a': 20,
}
```

```python
print(a)
```

```python
b = dict(
    c = 1,
    b = 'something',
    d = 20)
```

```python
print(b)
```

## Call value of an item based on its key

```python
a['b']
```

This method will product error when it can't find the selected key

```python
a['not known']
```

```python
a.get('b')
```

This method will not product an error when it can't find the selected key

```python
a.get('not known')
```

You can specify value when the selected key is not found

```python
a.get('not known', 'some value')
```

## Change value of an item

```python
a['c'] = 2000
```

```python
print(a)
```

## Update dictionary

```python
others = { 'Marx': 'Groucho', 'Howard': 'Moe' }
```

```python
a.update(others)
```

```python
print(a)
```

# 9. N-dimensional Array with Numpy

```python
import numpy as np
```

## One dimension array

```python
a = a = np.array([0, 1, 2, 3])
print(a)
print(a.shape)
print(a.ndim)
print(a.dtype)
```

## Two dimension array

```python
a = a = np.array([[0, 1, 2, 3],
                           [4, 5, 6, 7]])
print(a)
print(a.shape)
print(a.ndim)
print(a.dtype)
```

## Three dimension array

```python
a = np.array([[[0, 1, 2, 3],
                           [4, 5, 6, 7]],
                          [[0, 1, 2, 3],
                           [4, 5, 6, 7]],
                          [[0 ,1 ,2, 3],
                           [4, 5, 6, 7]]])

print(a)
print(a.shape)
print(a.ndim)
print(a.dtype)
```
