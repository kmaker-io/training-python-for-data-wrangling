## Small table performance comparison

```python
async def run_query():
    return await async_conn.fetch('select * from citibike')

# Now you can use a synchronous wrapper with %timeit
def sync_wrapper():
    return asyncio.get_event_loop().run_until_complete(run_query())
```

```python
%%timeit 
sync_wrapper()
```

```python
con = conn.raw_connection()
```

```python
import asyncio
import aiomysql
```

```python
async def async_mysql():
    async with aiomysql.create_pool(host='127.0.0.1', port=3306,
                                    user='root', password='airflow',
                                    db='data_wrangling') as pool:
        async with pool.acquire() as conn:
            async with conn.cursor() as cur:
                await cur.execute("select * from citibike_2019 where starttime like '2019-01%'")
                records = await cur.fetchall()
```

```python
def sync_wrapper():
    return asyncio.get_event_loop().run_until_complete(async_mysql())
```

```python
%%timeit
sync_wrapper()
```
